# Predictive Maintenance EDA
This notebook performs exploratory data analysis on the AI4I 2020 Predictive Maintenance Dataset.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os

df = pd.read_csv('dataset/predictive_maintenance.csv')
df.head()
```

```python
print(f'Dimensions: {df.shape}')
print('\nColumns:', df.columns.tolist())
print('\nData Types:')
print(df.dtypes)
```

```python
print('Missing Values:')
print(df.isnull().sum())
print('\nDuplicate Records:', df.duplicated().sum())
print('\nUnique Values per Column:')
print(df.nunique())
```

```python
print('Descriptive Statistics:')
display(df.describe())
```

```python
target = 'Target'
print('Target Distribution:')
print(df[target].value_counts())
fails = df[target].sum()
norms = len(df) - fails
print(f'\nNumber of machine failures: {fails}')
print(f'Number of normal records: {norms}')
print(f'Failure percentage: {(fails/len(df))*100:.2f}%')
```

```python
if 'Failure Type' in df.columns:
    print('\nAvailable Failure Categories:')
    print(df['Failure Type'].value_counts())
```

```python
import matplotlib.pyplot as plt
os.makedirs('results', exist_ok=True)

def plot_hist(col, bins=30):
    safe_name = col.replace(' ', '_').replace('[', '').replace(']', '')
    plt.figure(figsize=(8,5))
    plt.hist(df[col], bins=bins, edgecolor='k')
    plt.title(f'{col} Distribution')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.savefig(f'results/{safe_name}_dist.png')
    plt.show()

features_to_plot = ['Air temperature [K]', 'Process temperature [K]', 'Rotational speed [rpm]', 'Torque [Nm]', 'Tool wear [min]']
for feature in features_to_plot:
    if feature in df.columns:
        plot_hist(feature)
```

```python
if 'Target' in df.columns:
    plt.figure(figsize=(6,4))
    df['Target'].value_counts().plot(kind='bar', color=['blue', 'red'])
    plt.title('Machine Failure Distribution')
    plt.ylabel('Count')
    plt.xticks(ticks=[0,1], labels=['Normal (0)', 'Failure (1)'], rotation=0)
    plt.savefig('results/Machine_failure_dist.png')
    plt.show()
```

```python
if 'Failure Type' in df.columns:
    plt.figure(figsize=(8,5))
    df['Failure Type'].value_counts().plot(kind='bar', color='orange')
    plt.title('Failure Type Distribution')
    plt.ylabel('Count')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.savefig('results/Failure_type_dist.png')
    plt.show()
```

```python
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
corr = df[numeric_cols].corr()
plt.figure(figsize=(10,8))
plt.imshow(corr, cmap='coolwarm', vmin=-1, vmax=1)
plt.colorbar()
plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.savefig('results/Correlation_heatmap.png')
plt.show()
```
